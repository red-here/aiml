```python

rules = {
    "Fever and Cough": "Flu",
    "Flu": "Take Rest",
    "Rain": "Carry Umbrella"
}


```

```python
facts = ["Fever and Cough", "Rain"]


```

```python
def forward_chaining(facts):
    print("Forward Chaining:")
    for fact in facts:
        print("Fact:", fact)
        if fact in rules:
            print("Conclusion:", rules[fact])

```

```python

def backward_chaining(goal):
    print("\nBackward Chaining:")
    if goal in rules.values():
        for condition, result in rules.items():
            if result == goal:
                print("Goal:", goal)
                print("Rule:", condition, "->", goal)
                return
    print("Goal cannot be proved.")
```

```python


forward_chaining(facts)
backward_chaining("Flu")
```
