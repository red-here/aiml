```python
# Graph with edge costs
graph = {
    'A': {'B': 2, 'C': 1},
    'B': {'G': 2},
    'C': {'G': 10},
    'G': {}
}

# Heuristic values
h = {'A': 2, 'B': 2, 'C': 1, 'G': 0}

start = 'A'
goal = 'G'
```

```python
# GREEDY BEST FIRST SEARCH

def greedy():

    open_list = [start]
    visited = []
    parent = {}

    while open_list:

        # Select node with minimum heuristic
        current = open_list[0]

        for node in open_list:
            if h[node] < h[current]:
                current = node

        open_list.remove(current)

        if current == goal:
            break

        visited.append(current)

        # Add neighbouring nodes
        for neighbour in graph[current]:

            if neighbour not in visited and neighbour not in open_list:
                open_list.append(neighbour)
                parent[neighbour] = current

    # Display path
    if current == goal:

        path = [goal]

        while path[-1] != start:
            path.append(parent[path[-1]])

        path.reverse()

        cost = 0
        for i in range(len(path) - 1):
            cost += graph[path[i]][path[i + 1]]

        print("Greedy Path:", path)
        print("Cost:", cost)

    else:
        print("Path not found")
```

```python

# A STAR SEARCH

def astar():

    open_list = [start]
    visited = []
    parent = {}
    g = {start: 0}

    while open_list:

        # Select node with minimum g + h
        current = open_list[0]

        for node in open_list:
            if g[node] + h[node] < g[current] + h[current]:
                current = node

        open_list.remove(current)

        if current == goal:
            break

        visited.append(current)

        # Add neighbouring nodes
        for neighbour in graph[current]:

            new_cost = g[current] + graph[current][neighbour]

            if neighbour not in g or new_cost < g[neighbour]:

                g[neighbour] = new_cost
                parent[neighbour] = current

                if neighbour not in open_list:
                    open_list.append(neighbour)

    # Display path
    if current == goal:

        path = [goal]

        while path[-1] != start:
            path.append(parent[path[-1]])

        path.reverse()

        print("A* Path:", path)
        print("Cost:", g[goal])

    else:
        print("Path not found")
```

```python
print("--- GREEDY SEARCH ---")
greedy()

print("\n--- A* SEARCH ---")
astar()
```
