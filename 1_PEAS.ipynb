```python

class SmartHomeAgent:
    def __init__(self):
        self.light = False
        self.fan = False
        self.ac = False

    def act(self, command, temperature):
        command = command.lower()

        if command == "light on":
            self.light = True
            print("Light is ON.")

        elif command == "light off":
            self.light = False
            print("Light is OFF.")

        elif command == "fan on":
            self.fan = True
            print("Fan is ON.")

        elif command == "fan off":
            self.fan = False
            print("Fan is OFF.")

        elif command == "ac on":
            self.ac = True
            print("Air Conditioner is ON.")

        elif command == "ac off":
            self.ac = False
            print("Air Conditioner is OFF.")

        elif command == "auto":
            if temperature > 30:
                self.ac = True
                self.fan = True
                print("Room is hot. AC and Fan turned ON.")
            elif temperature > 25:
                self.fan = True
                self.ac = False
                print("Room is warm. Fan turned ON.")
            else:
                self.fan = False
                self.ac = False
                print("Room temperature is comfortable. Fan and AC turned OFF.")

        elif command == "status":
            print("Light:", "ON" if self.light else "OFF")
            print("Fan:", "ON" if self.fan else "OFF")
            print("AC:", "ON" if self.ac else "OFF")

        else:
            print("Invalid command. Please try again.")


agent = SmartHomeAgent()

print("SMART HOME INTELLIGENT AGENT")
print("Commands: light on/off, fan on/off, ac on/off, auto, status, exit")

while True:
    command = input("\nEnter command: ").lower()

    if command == "exit":
        print("Agent stopped.")
        break

    if command == "auto":
        temperature = float(input("Enter room temperature (°C): "))
    else:
        temperature = 0

    agent.act(command, temperature)

```
